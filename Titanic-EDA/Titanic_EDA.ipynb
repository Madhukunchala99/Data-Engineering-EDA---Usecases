# Data Cleaning

Before analyzing the dataset, it is important to clean the data. This involves checking for missing values, duplicate records, incorrect data types, and other inconsistencies that may affect the analysis.

## Checking Missing Values

Missing values occur when information is not available for some records. Identifying these values helps us decide whether to remove or replace them.

```python
df.isnull()
```

```python
df.isnull().sum()
```

### Observation

From the output, we can identify which columns contain missing values.

Columns with missing values require appropriate handling before further analysis.

For example:

- Age contains missing values.
- Cabin contains a large number of missing values.
- Embarked contains only a few missing values.

## Checking Duplicate Records

Duplicate records can affect the accuracy of the analysis. Therefore, we check whether the dataset contains any duplicate rows.

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

### Observation

The output shows the total number of duplicate records in the dataset.

If duplicate rows exist, they should be removed to avoid biased analysis.

## Handling Missing Values

Different techniques can be used to handle missing values depending on the type of data.

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
```

```python
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
```

```python
df.drop(columns=["Cabin"], inplace=True)
```

### Observation

- Missing values in the **Age** column were filled using the median.
- Missing values in the **Embarked** column were filled using the mode.
- The **Cabin** column was removed because it contained too many missing values.

```python
df.isnull().sum()
```

### Observation

After handling the missing values, the dataset no longer contains missing values in the columns used for analysis.

The dataset is now cleaner and ready for exploratory data analysis.

```python
df.duplicated().sum()
```

### Observation

Duplicate records have been removed successfully.

The dataset is now free from duplicate rows and is ready for further analysis.

# Multivariate Analysis

Multivariate Analysis examines the relationship among three or more variables simultaneously. It helps uncover hidden patterns and provides a deeper understanding of the dataset.

## Gender, Passenger Class and Survival

This analysis examines how gender and passenger class together influenced survival.

```python
sns.catplot(
    data=df,
    x="Pclass",
    hue="Survived",
    col="Sex",
    kind="count",
    height=5,
    aspect=1
)

plt.show()
```

### Observation

Compare the survival counts for males and females across all passenger classes.

Identify which combination had the highest survival rate.

## Age, Fare and Survival

This scatter plot shows the relationship between Age and Fare while highlighting passenger survival.

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare by Survival")
plt.show()
```

### Observation

Observe whether passengers with higher fares had better survival rates.

## Pair Plot

A Pair Plot displays relationships among all numerical variables simultaneously.

```python
sns.pairplot(
    df[
        ["Age",
         "Fare",
         "Pclass",
         "SibSp",
         "Parch",
         "Survived"]
    ],
    hue="Survived"
)

plt.show()
```

### Observation

The pair plot helps identify relationships between multiple numerical variables and the survival status.

## Correlation Analysis

Correlation measures the strength of the relationship between numerical variables.

Values range from:

- +1 → Strong Positive Correlation
- 0 → No Correlation
- -1 → Strong Negative Correlation

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

Identify the variables with the strongest positive and negative correlations.

Discuss whether Age, Fare, Passenger Class, or Family-related features are associated with survival.

```python
sns.barplot(
    data=df,
    x="Embarked",
    y="Survived"
)

plt.title("Average Survival Rate by Embarkation Port")

plt.show()
```
